# 1. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# 2. Load Dataset

```python
# Load the datasets
data_path = "../data/Faults.NNA"
columns_path = "../data/Faults27x7_var"

# Read column names
with open(columns_path, "r") as f:
    columns = [line.strip() for line in f]

# Read the dataset
df = pd.read_csv(
    data_path,
    sep="\t",
    header=None,
    names=columns
)

print("Dataset loaded successfully!")
```

# 3. Basic Dataset Inspection

```python
print("Dataset Shape:", df.shape)

print("\nFirst 5 Rows:")
display(df.head())

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

## 4.5 Defect Class Distribution

```python
fault_columns = [
    "Pastry",
    "Z_Scratch",
    "K_Scatch",
    "Stains",
    "Dirtiness",
    "Bumps",
    "Other_Faults"
]

fault_counts = df[fault_columns].sum()

print("Number of plates in each defect class:")
display(fault_counts)
```

```python
# Verify defect labels

labels_per_row = df[fault_columns].sum(axis=1)

print("Minimum labels per row:", labels_per_row.min())
print("Maximum labels per row:", labels_per_row.max())

print("\nRows with exactly one defect label:",
      (labels_per_row == 1).sum())

print("Total rows:", len(df))
```

## 4.6 Target Variable

```python
fault_columns = [
    "Pastry",
    "Z_Scratch",
    "K_Scatch",
    "Stains",
    "Dirtiness",
    "Bumps",
    "Other_Faults"
]

df["Defect_Type"] = df[fault_columns].idxmax(axis=1)

display(df[["Defect_Type"]].head())
```

```python
# Verify Target Variable

print("Defect Type Distribution:")
display(df["Defect_Type"].value_counts())
```
