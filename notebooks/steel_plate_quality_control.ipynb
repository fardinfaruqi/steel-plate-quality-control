## 6.7 Outlier and Relationship Analysis

### Step 1: Calculate outliers using IQR

```python
# Detect outliers using the IQR method

outlier_summary = []

for feature in X.columns:
    Q1 = X[feature].quantile(0.25)
    Q3 = X[feature].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = X[
        (X[feature] < lower_bound) |
        (X[feature] > upper_bound)
    ]

    outlier_summary.append({
        "Feature": feature,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": len(outliers)
    })

outlier_summary = pd.DataFrame(outlier_summary)

display(
    outlier_summary
    .sort_values("Outlier Count", ascending=False)
)
```

### Step 2: Visualize the number of outliers

```python
# Visualize outlier counts

outlier_plot = outlier_summary.sort_values(
    "Outlier Count",
    ascending=False
)

plt.figure(figsize=(12, 7))

sns.barplot(
    data=outlier_plot,
    x="Outlier Count",
    y="Feature"
)

plt.title("Number of Potential Outliers by Feature")
plt.xlabel("Number of Potential Outliers")
plt.ylabel("Feature")

plt.show()
```

### Step 3: Look at the strongest feature relationships

```python
# Find the strongest feature relationships

correlation_pairs = correlation_matrix.where(
    np.triu(np.ones(correlation_matrix.shape), k=1).astype(bool)
)

correlation_pairs = (
    correlation_pairs
    .stack()
    .reset_index()
)

correlation_pairs.columns = ["Feature 1", "Feature 2", "Correlation"]

correlation_pairs["Absolute Correlation"] = (
    correlation_pairs["Correlation"].abs()
)

strongest_relationships = correlation_pairs.sort_values(
    "Absolute Correlation",
    ascending=False
)

display(strongest_relationships.head(10))
```

## 7.7 SVM Hyperparameter Tuning

```python
# SVM Hyperparameter Tuning

from sklearn.model_selection import GridSearchCV

# Parameters to test
param_grid = {
    "C": [0.1, 1, 10],
    "kernel": ["linear", "rbf"]
}

print("Parameter combinations:")
print(param_grid)
```

### 7.7.2 Run Grid Search

```python
# Create the base SVM

svm_base = SVC(
    random_state=42
)

# Create Grid Search
grid_search = GridSearchCV(
    estimator=svm_base,
    param_grid=param_grid,
    cv=5,
    scoring="f1_weighted",
    n_jobs=-1
)

# Perform hyperparameter tuning
grid_search.fit(X_train_scaled, y_train)

print("Hyperparameter tuning completed.")
```

```python
# Display Grid Search results

grid_results = pd.DataFrame(grid_search.cv_results_)

grid_results = grid_results[
    [
        "param_C",
        "param_kernel",
        "mean_test_score",
        "std_test_score",
        "rank_test_score"
    ]
].sort_values("rank_test_score")

display(grid_results)
```

### 7.7.5 Get the Best SVM

# 9. Outlier Detection

## 9.1 Z-Score Method

### 9.1.1 Calculate Z-Scores

```python
# Z-Score Method

from scipy.stats import zscore

# Calculate Z-scores for all features
z_scores = np.abs(zscore(X_cluster_scaled))

print("Z-score matrix shape:", z_scores.shape)
```

### 9.1.2 Identify Outlier Observations

```python
# Identify observations with at least one extreme Z-score

z_threshold = 3

z_outlier_mask = (z_scores > z_threshold).any(axis=1)

print("Z-score threshold:", z_threshold)
print("Number of potential outlier observations:", z_outlier_mask.sum())
print("Percentage of potential outliers:",
      round(z_outlier_mask.mean() * 100, 2), "%")
```

### 9.1.3 Count Extreme Features Per Observation

```python
# Count the number of extreme features for each observation

extreme_feature_count = (z_scores > z_threshold).sum(axis=1)

print("Maximum number of extreme features in one observation:",
      extreme_feature_count.max())

print("Observations with at least one extreme feature:",
      (extreme_feature_count > 0).sum())
```

### 9.1.4 Create an Outlier Summary

```python
# Create Z-score outlier summary

z_outlier_summary = pd.DataFrame({
    "Z_Outlier": z_outlier_mask,
    "Extreme_Feature_Count": extreme_feature_count
})

display(z_outlier_summary.head())
```

### 9.1.5 Find the Most Extreme Observations

```python
# Find observations with the highest number of extreme features

top_z_outliers = z_outlier_summary.sort_values(
    by="Extreme_Feature_Count",
    ascending=False
).head(10)

display(top_z_outliers)
```

### 9.1.6 Connect Z-Score Outliers to Defect Types

```python
# Add defect type to the Z-score outlier summary

z_outlier_summary["Defect_Type"] = y.values

display(
    z_outlier_summary[
        z_outlier_summary["Z_Outlier"] == True
    ]["Defect_Type"].value_counts()
)
```

### 9.1.7 Visualize Extreme Feature Counts

```python
# Distribution of extreme feature counts

plt.figure(figsize=(8, 5))

plt.hist(
    extreme_feature_count,
    bins=range(
        int(extreme_feature_count.max()) + 2
    ),
    edgecolor="black"
)

plt.title("Distribution of Extreme Feature Counts")
plt.xlabel("Number of Features with |Z| > 3")
plt.ylabel("Number of Observations")

plt.show()
```

## 9.2 IQR Method

### 9.2.1 Calculate Q1, Q3 and IQR

```python
# 9.2 IQR Method

# Calculate first and third quartiles
Q1 = X.quantile(0.25)
Q3 = X.quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

print("IQR calculation completed.")
```

### 9.2.2 Calculate Lower and Upper Bounds

```python
# Calculate IQR-based lower and upper bounds

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower and upper bounds calculated.")
```

### 9.2.3 Identify Potential Outliers

```python
# Identify observations with at least one IQR-based outlier feature

iqr_outlier_mask = (
    (X < lower_bound) |
    (X > upper_bound)
).any(axis=1)

print(
    "Number of potential IQR outlier observations:",
    iqr_outlier_mask.sum()
)

print(
    "Percentage of potential IQR outliers:",
    round(iqr_outlier_mask.mean() * 100, 2),
    "%"
)
```

### 9.2.4 Count Extreme Features

```python
# Count the number of IQR-outlier features per observation

iqr_extreme_feature_count = (
    (X < lower_bound) |
    (X > upper_bound)
).sum(axis=1)

print(
    "Maximum number of IQR-outlier features:",
    iqr_extreme_feature_count.max()
)

print(
    "Observations with at least one IQR outlier feature:",
    (iqr_extreme_feature_count > 0).sum()
)
```

### 9.2.5 Create IQR Outlier Summary

```python
# Create IQR outlier summary

iqr_outlier_summary = pd.DataFrame({
    "IQR_Outlier": iqr_outlier_mask,
    "Extreme_Feature_Count": iqr_extreme_feature_count
})

display(iqr_outlier_summary.head())
```
