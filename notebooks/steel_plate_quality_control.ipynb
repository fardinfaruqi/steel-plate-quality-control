# 1. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# 2. Load Dataset

```python
# Load the datasets
data_path = "../data/Faults.NNA"
columns_path = "../data/Faults27x7_var"

# Read column names
with open(columns_path, "r") as f:
    columns = [line.strip() for line in f]

# Read the dataset
df = pd.read_csv(
    data_path,
    sep="\t",
    header=None,
    names=columns
)

print("Dataset loaded successfully!")
```

# 3. Basic Dataset Inspection

```python
print("Dataset Shape:", df.shape)

print("\nFirst 5 Rows:")
display(df.head())

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

## 4.5 Defect Class Distribution

```python
fault_columns = [
    "Pastry",
    "Z_Scratch",
    "K_Scatch",
    "Stains",
    "Dirtiness",
    "Bumps",
    "Other_Faults"
]

fault_counts = df[fault_columns].sum()

print("Number of plates in each defect class:")
display(fault_counts)
```

```python
# Verify defect labels

labels_per_row = df[fault_columns].sum(axis=1)

print("Minimum labels per row:", labels_per_row.min())
print("Maximum labels per row:", labels_per_row.max())

print("\nRows with exactly one defect label:",
      (labels_per_row == 1).sum())

print("Total rows:", len(df))
```

## 4.6 Target Variable

```python
fault_columns = [
    "Pastry",
    "Z_Scratch",
    "K_Scatch",
    "Stains",
    "Dirtiness",
    "Bumps",
    "Other_Faults"
]

df["Defect_Type"] = df[fault_columns].idxmax(axis=1)

# Remove the original fault indicator columns
df = df.drop(columns=fault_columns)

print("New Dataset Shape:", df.shape)
```

```python
# Verify Target Variable

print("Defect Type Distribution:")
display(df["Defect_Type"].value_counts())
```

# 5. Data Preprocessing

## 5.1 Separate Features and Target

```python
X = df.drop(columns=["Defect_Type"])
y = df["Defect_Type"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)

print("\nFeatures:")
print(X.columns.tolist())

print("\nTarget:")
print(y.name)
```

## 5.2 Check Feature Scales

```python
feature_summary = pd.DataFrame({
    "Minimum": X.min(),
    "Maximum": X.max(),
    "Mean": X.mean(),
    "Standard Deviation": X.std()
})

display(feature_summary)
```

## 5.3 Feature Scaling - Standardization

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

X_scaled = pd.DataFrame(
    X_scaled,
    columns=X.columns,
    index=X.index
)

display(X_scaled.head())
```

```python
# Verify Standardization

print("Mean after standardization:")
display(X_scaled.mean().round(2))

print("\nStandard deviation after standardization:")
display(X_scaled.std().round(2))
```
